You are given an array of distinct integers nums, sorted in ascending order, and an integer target.

Implement a function to search for target within nums. If it exists, then return its index, otherwise, return -1.

Your solution must run in O(logn) time.

Example 1:

Input: nums = [-1,0,2,4,6,8], target = 4

Output: 3


Input: nums = [-1,0,2,4,6,8], target = 3

Output: -1

```python
def search(nums,target):
    L,R=0,len(nums)-1

    while L<=R:
        mid=(L+R)//2
        if target>nums[mid]:
            L=mid+1
        elif target<nums[mid]:
            R=mid-1
        else:
            return mid
    return -1

RESULT=search( nums = [-1,0,2,4,6,8],target=3)

print(RESULT)
```

You are given an m x n 2-D integer array matrix and an integer target.

Each row in matrix is sorted in non-decreasing order.The first integer of every row is greater than the last integer of the previous row.

Return true if target exists within matrix or false otherwise.

Can you write a solution that runs in O(log(m * n)) time?


Input: matrix = [[1,2,4,8],[10,11,12,13],[14,20,30,40]], target = 10

Output: true

The entire matrix is sorted in ascending order, so we can use binary search to find a target value efficiently.

If we consider a normal sorted array containing n elements, we can find a target using binary search in O(log n) time. Here we have 'm' different rows. if we run a binary search on every single row until we find the target value, then the time complexity will be "O(m logn).".

We can do this in a better way since the first integer of every row is greater than the last integer of the previous row.ie each row itself in a sorted order. Instead of searching every row individually, we can perform a binary search on the rows to identify which row could contain the target. 

the top row will be having smaller values than bottom row.Hence binarysearch will figureout which one of these "m" rows we are going to search for the target.Since there are m rows, this step takes O(log m) time.
 
Once we identify the appropriate row, we perform a second binary search within that row. Since each row contains n elements, this step takes O(log n) time.

so our total time complexity becomes " O(log m+log n) ". 

The extra space complexity is O(1) because we only use a few variables to track the search boundaries and middle positions.




```python
'''
We're using binary search twice:

1. Find which row could contain the target.
2. Search inside that row.'''

class SolutionMatrix:
    def searchMatrix(self, matrix, target):
        ROWS,COLS=len(matrix),len(matrix[0])
        top,bottom=0,ROWS-1

        while top<=bottom:
            middleRow=(top+bottom)//2
            if target >matrix[middleRow][-1]: #If the target value is greater than the last value of the middle row, move the top pointer down
                top=middleRow+1 #ie target must be in a lower row, if it exists.
            elif target <matrix[middleRow][0]:  #if target value is less than the first value of the middlerow,move the bottom pointer  up
                bottom=middleRow -1
            else:
                break
        if not (top <= bottom):
            return False


        #middleRow=(top+bottom)//2
        l,r=0,COLS-1

        while l <= r :
            m=(l+r)//2
            if target > matrix[middleRow][m]: # if the target is greater than the middle value of middlerow then move the 'l' pointer to right
                l=m+1
            elif target <matrix[middleRow][m]:
                r=m -1
            else:
                return True # when both previous conditions are false the target must equal the middle element.
        return False
    
solutionmatrix=SolutionMatrix()

result = solutionmatrix.searchMatrix(
    matrix=[[1, 2, 4, 8], [10, 11, 12, 13], [14, 20, 30, 40]],
    target=10
)
print(result)
        
```
