# Max Consecutive Ones


You are given a binary array nums, return the maximum number of consecutive 1's in the array.

Example 1:

Input: nums = [1,1,0,1,1,1]

Output: 3

### 1. Brute Force

```python
class Solution:
    #def findMaxConsecutiveOnes(self, nums: List[int]) -> int: 
    '''List[int] is a type hint. ie,nums is expected to be a list containing integers.
       -> int.     means-The function is expected to return an integer.'''
    
    def findMaxConsecutiveOnes(self, nums):
        n, res = len(nums), 0

        for i in range(n):
            cnt = 0
            for j in range(i, n):#i=0 ,#j=0 cnt=1,j=1 cnt=2 ,j=2 exit ,res=max(0,2)=2 #i=1 cnt=0 ,j=1 cnt=1,j=2 exit res=max(2,1)=2,#i=2 j=2 cnt=0 res(2,0)=2,
                                  #i=3 ,j=3 ,cnt=1 ,j=4 cnt=2 ,j=5 cnt=3 res=max(2,3)=3 ,#i=4 j=4 cnt=1 j=5 cnt=2 res=max(3,2)=3 #i=5 j=5 cnt=1 res=max(3,1)=3
                if nums[j] == 0:
                    break
                cnt =cnt+1 
            res = max(res, cnt)

        return res
    
solution=Solution()
result=solution.findMaxConsecutiveOnes(nums=[1,1,0,1,1,1])
print(result)


#However, it uses two loops, so its time complexity is O(n²) in the worst case
```

```python
def findmaxrep(nums):

    res = 0
    count = 0

    for num in nums: 

        if num == 1:   #1 cnt=1 res=(0,1)=1, #1 cnt=2 res=(1,2)=2, #0 cnt=0(reset) res=(2,0)=2 ,#1 cnt=1 res=(2,1),#1 cnt=2 res(2,2)=2 ,#1 cnt=3 res=(2,3)=3
            count += 1
        else:
            count = 0    #cnt = cnt + 1 if num else 0

        res = max(res, count)

    return res
findmaxrep([1, 1, 0, 1, 1, 1])

#time complexity Time complexity: O(n)
```

# Concatenation of Array

you are given an integer array nums of length n. Create an array ans of length 2n where ans[i] == nums[i] and ans[i + n] == nums[i] for 0 <= i < n (0-indexed). Specifically, ans is the concatenation of two nums arrays.Return the array ans.

Example 1:

Input: nums = [1,4,1,2]

Output: [1,4,1,2,1,4,1,2]

```python
def getConcatenation(nums):
    n=len(nums)
    ans = [0] * (2*n)
    for i in range(n):
        
        ans[i] = nums[i] 
        ans[i+n]=nums[i]
        print(f"ans is {ans}")
      
    return ans

getConcatenation([1,4,5,2])
```

```python
def getConcatenation(nums):
    n=len(nums)
    ans = [0] * (2*n)
    for i, num in enumerate(nums):
            ans[i] = ans[i + n] = num
    return ans

getConcatenation([1,4,5,2])
```

# Design a Dynamic Array

Design a Dynamic Array (aka a resizable array) class, such as an ArrayList in Java or a vector in C++.




length   = how many elements are currently stored

capacity = how much space has been allocated

length tells us how much data we have; capacity tells us how much storage we have.And when length == capacity, we double the capacity, copy the existing elements, and continue adding.

```python
# Dynamic Array implementation
# Note: Python lists are dynamic arrays by default,
# but this is an example of what's going on under the hood.
class DynamicArray:
    
    def __init__(self, capacity: int):
        self.capacity = capacity
        self.length = 0
        self.arr = [0] * self.capacity

    # Get value at i-th index
    def get(self, i: int) -> int:
        return self.arr[i]

    # Set n at i-th index
    def set(self, i: int, n: int) -> None: # inserting
        self.arr[i] = n

    # #Add a new element 'n' to the end of the array. If the array is full, resize first.
    def pushback(self, n: int) -> None:
        if self.length == self.capacity:
            self.resize()
            
        # insert at next empty position
        self.arr[self.length] = n #suppose arr=[10, 20, 0, 0],length=2, capacity=4,the next empty position is index 2 since length = 2.  self.arr[2] = 30, arr becomes [10, 20, 30, 0]
        self.length += 1          # before length=2 after inserting 30 , length =3. 

    # Remove the last element in the array
    def popback(self) -> int:
        if self.length > 0:
            # soft delete the last element
            self.length -= 1  #suppose length = 3 ,after operation it becomes length=2 and it return self.arr[2] which is the element we just removed.
        # return the popped element
        return self.arr[self.length]

    def resize(self) -> None:
        # Create new array of double capacity
        self.capacity = 2 * self.capacity #  #suppose capacity = 4 , #after doubling capacity = 8
        new_arr = [0] * self.capacity   #Create a new array [0, 0, 0, 0, 0, 0, 0, 0]
        
        # Copy the existing elements to new_arr
        for i in range(self.length):
            new_arr[i] = self.arr[i]
        self.arr = new_arr # Replace the old array #befor  self.arr →[10, 20, 30, 40] #after →[10, 20, 30, 40, 0, 0, 0, 0]

    def getSize(self) -> int:
        return self.length
    
    def getCapacity(self) -> int:
        return self.capacity

 #supose arr = [10, 20, 30, 0, 0] ,length = 3 ,The actual elements occupy: index 0 → 10 index 1 → 20 ,index 2 → 30
    #So the next available position is exactly: index 3  which is self.length . therefore self.arr[self.length] = n
```
