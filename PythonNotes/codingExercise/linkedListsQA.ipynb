# Reverse Linked List
Given the beginning of a singly linked list head, reverse the list, and return the new beginning of the list.

Example 1:

Input: head = [0,1,2,3]

Output: [3,2,1,0]

```python
''' since we are coding in vscode/python jupiter notebook ,we need to convert the input into nodes first.
If you're solving this on LeetCode/NeetCode, you normally do not give the input yourself. The platform converts:head = [0,1,2,3]
'''

class ListNode:
    def __init__(self, data):
        self.data = data
        self.next = None


node1 = ListNode(0)
node2 = ListNode(1)
node3 = ListNode(2)
node4 = ListNode(3)

node1.next=node2
node2.next=node3
node3.next=node4

head=node1  
```

```python
class Solution:
    def reverseList(self, head):

        prev = None 
        curr = head # ie head = 0 , so curr=0  head[0] → [1] → [2] → [3] → None

        while curr != None:

            temp = curr.next # curr=0 curr.next=1 ie temp=1
            curr.next = prev  #  Make curr.next point to whatever prev is pointing to. prev=none and 0.next=1 ,so now move 0.next=none ,0 → None , 1(temp) → 2 → 3 → None

            prev = curr  #  crr=0 so prev=0
            curr = temp # temp=1 so curr=1   ie prev=0 curr=1 

        return prev #When the loop ends, prev is the new head of the reversed list.


'''
While curr exists:
Save the next node: temp = curr.next
Reverse the pointer: curr.next = prev
Move prev to curr
Move curr to temp '''
```

```python
# -------- Call the function --------

solution = Solution()

new_head = solution.reverseList(head)


# -------- Print the result --------

cur = new_head

while cur != None:
    print(cur.data, end=" ")
    cur = cur.next
```
