A linked list is a linear data structure where elements are not stored in contiguous (neighboring) memory locations.

The individual items are called nodes and connected with each other using links. 

Nodes are stored in random order in memory.

Each node contains two things: 
1. data → the actual value ,It could be a character, an integer, etc
2. next → a reference/pointer to the next node (The next reference tells us where the next node is)






# Creating a Linked List from scratch



```python
class ListNode:
    def __init__(self, data):
        self.data = data
        self.next = None

# # Suppose that we have 3 ListNode objects – Node1, Node2, Node3
node1 = ListNode(10)
node2 = ListNode(20)
node3 = ListNode(50)

node1.next = node2
node2.next = node3
node3.next= None #Only the last node in the linked list would have its next pointer point to null

#Linked Lists have a head, and a tail pointer.   
head=node1   #linked list doesn't have indexes like an array.We need to know Where does my linked list begin. That's the purpose of head.

tail=node3 #tail pointer points to the very last node

''' If there is only one node in the Linked List, the head and the tail point to the same node.'''
```

# Traversing a Linked List

```python
cur=head
while(cur!=None):
    print(cur.data)
    cur = cur.next # move to next node
```

1. We start the traversal at the head of the list, which is ListNode1.

2. We assign it to a variable cur, denoting the current node we are at.

3. We execute the while loop until we reach the end of the list which is null.

4. In each iteration, we update cur to be the next node in the list by setting cur = cur.next.

5. The traversal runs in O(n) time where n is the number of nodes in the linked list.

# Circular Linked List

if ListNode3’s next pointer is set to ListNode1 instead of null. This results in a circular linked list.Attempting to iterate through a circular linked list would result in an infinite loop. We would never reach the end of the linked list.

# Operations of a Singly Linked List

## Appending

An advantage that Linked Lists have over arrays is that inserting a new element can be performed in O(1) time, even if we insert in the middle.

We do not have to shift any elements since there is no requirement for the elements to be stored contiguously in memory.

This assumes we already have a reference to the node at the desired position we want to insert. If we have to traverse the list to arrive at the insertion point, the operation would take O(n) time.




```python
#We want to insert 30 between 20 and 50.     ie 10 20 50

node4 = ListNode(30)
node4.next = node2.next 
node2.next = node4


print("\nAfter inserting 30:")
cur = head
while cur != None:
    print(cur.data)
    cur = cur.next

```

```python
    
'''If we wanted to append a ListNode4 to the end of the list, we would be appending to the tail. 

Once ListNode4 is appended, we update our tail pointer to be at ListNode4. This operation would be done in O(1) time since it is only one operation.'''


node5 = ListNode(60)
tail.next=node5  ## Connect old tail (50) to 60
tail=node5 ## Move tail to the new last node


print("\nAfter adding 60 at the end:")
cur = head
while cur != None:
    print(cur.data)
    cur = cur.next
```

## Deleting from a Singly Linked List

Deleting a node from a singly linked list will take O(1) since we can accomplish this by updating a single pointer.

This assumes we already have a reference to the node at the desired position we want to delete. If we have to traverse the list to arrive at the deletion point, the operation would take O(n) time.

```python
'''Suppose we want to delete ListNode2 

Currently, our head points to ListNode1, and head.next points to ListNode2.    cur.next=cur.next.next
We can update our head.next pointer to ListNode3, which can be accessed by chaining next pointers like head.next.next

This makes sense since head.next is ListNode2, and logically, head.next.next would be ListNode3.
'''

head.next = head.next.next #ListNode1’s(head) next pointer points to ListNode3, instead of ListNode2

print("\nAfter removing node 2 ie 20:")
cur = head
while cur != None:
    print(cur.data)
    cur = cur.next


'''It can be assumed that the memory occupied by ListNode2 will be cleared via garbage collection in most languages.'''
```

ie, why we use Linked Lists? - Because insertion and deletion can be very efficient when you already have the correct node/location.

# Doubly linked lists

As the name implies, each node now has two pointers. In addition to the next pointer, we have a prev pointer which points to the previous node.

If the prev pointer points to null, it is an indication that we are at the head of the linked list.

# Queues

A Queue Data Structure follows the principle of "First in, First out" (FIFO), where the first item added to the queue is the first one to be removed.

The easiest way to implement a queue is using a linked list.

The main two operations that queues support are enqueue and dequeue.
