# Binary Tree

1. Nodes in a binary tree also have at most 2 pointers,  we call them the 'left child' and the 'right child' pointers.

2. The first node in a binary tree is referred to as the 'root node'
3. If a node does not have any children, it is classified as a 'leaf node'
4. Unlike linked lists, binary tree node pointers can only point in one direction. As such, cycles are not allowed in binary trees
5. The 'height' of a binary tree is measured from the root node all to way to the lowest leaf node
6. The height of a single node tree is just 1, if the node itself is counted, or 0 if not.
7. 'Depth' of a binary tree node is measured from itself all the way up to the root.the depth at the root node is 1, with it increasing as we go down.
8. A node connected to all of the nodes below it is considered an 'ancestor' to those nodes.root node is an ancestor to all of the nodes in the tree.
9. The 'descendent' of a node is either child of the node or child of some other descendent of the node.

```python
class TreeNode:
    def __init__(self, val):
        self.val = val
        self.left = None
        self.right = None
```

# Binary Search Trees

1. Binary Search Trees (BST) are a variation of binary trees with the addition of a sorted property. 

2. The property is that every node in the left subtree is smaller than the root and every node in the right subtree is greater than the root.

3. BSTs are often preferred over sorted arrays because they allow for insertion and deletion in O(log n) time.

4. If the target is greater than the current node, we search the right subtree.

5. If we have a balanced binary tree, our search algorithm will run in O(log n) time. Balanced binary tree means that the height of the left-subtree is equal to the height of the right-subtree, or there is a difference of 1.

In a balanced tree, we can eliminate half the nodes each time, which results in O(log n).

In the worst case we have a skewed tree, where the height of the tree is equal to the number of nodes. In this case, the time complexity will be O(n).

```python
def search(root, target):
    if not root:
        return False
    
    if target > root.val:
        return search(root.right, target)
    elif target < root.val:
        return search(root.left, target)
    else:
        return True
```

## insertion

```python
# Insert a new node and return the root of the BST.
def insert(root, val):
    if not root:
        return TreeNode(val) #found an empty position. Create and return the new node.
    
    if val > root.val:
        root.right = insert(root.right, val) 
    elif val < root.val:
        root.left = insert(root.left, val)
    return root
```

## deletion

```python
# Return the minimum value node of the BST.
def minValueNode(root):
    curr = root
    while curr and curr.left:
        curr = curr.left
    return curr

# Remove a node and return the root of the BST.
def remove(root, val):
    if not root:
        return None
    
    if val > root.val:
        root.right = remove(root.right, val)
    elif val < root.val:
        root.left = remove(root.left, val)

    else: #Case 2: Node has one child
        if not root.left:
            return root.right #No left child? Return the right child.
        elif not root.right:
            return root.left

        else:#Case 3: Node has two children
            minNode = minValueNode(root.right)
            root.val = minNode.val
            root.right = remove(root.right, minNode.val)
    return root
```

## BST traversal


Traversal means visiting every node in a particular order.

## depth-first search.
The idea is we pick a direction, say left, and keep following pointers as far down left as we can go until we reach null. Once we reach null, we backtrack to the parent node and then go right. We keep doing this until we have visited every node in the tree. This is the essence of depth-first search.

There are three ways to traverse a tree using depth-first search:

Inorder

Preorder

Postorder


### Inorder Traversal

An inorder traversal will recursively visit all the nodes in the left subtree, then visit the parent node and finally visit all the nodes in the right subtree. In this case, "visit" could mean anything from printing the node to performing some operation on it.

It is important to note that an inorder traversal will only print the nodes in a sorted order if the tree is a binary search tree.





```python
#inorder implementation
def inorder(root):
    if root is None:
        return

    inorder(root.left)
    print(root.val, end=" ")
    inorder(root.right)
```

### Preorder Traversal

preorder traversal will visit the parent node first, then visit the left subtree and finally visit the right subtree.

```python
#pre order

def preorder(root):
    if not root:
        return    
    print(root.val)
    preorder(root.left)
    preorder(root.right)
```

### Postorder Traversal

A postorder traversal will visit the left subtree, then the right subtree and finally the parent node last.


```python
def postorder(root):
    if not root:
        return    
    postorder(root.left)
    postorder(root.right)
    print(root.val)
```
